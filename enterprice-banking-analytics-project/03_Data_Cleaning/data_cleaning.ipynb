```python
# import essential libraries

import numpy as np
import pandas as pd
```

```python
# load addresses.csv file

addresses_df=pd.read_csv("C:/Users/Sistech Computer/OneDrive/Desktop/sandeep/enterprice-banking-analytics-system/data/raw/addresses.csv")
addresses_df.head()
```

```python
# check the size of addresses.csv file
addresses_df.shape
```

```python
# check datatypes of addresses.csv
addresses_df.dtypes
```

```python
# print last row
addresses_df.tail()
```

```python
# describe the dataset
addresses_df.describe()
```

```python
# check the missing values
addresses_df.isnull().sum()
```

```python
# fill the missing values
addresses_df['Street']=addresses_df['Street'].fillna('Unknown')
addresses_df['City']=addresses_df['City'].fillna(addresses_df['City'].mode()[0])
addresses_df['Country']=addresses_df['Country'].ffill()
```

```python
# check duplicates values
addresses_df.duplicated().sum()
```

```python
# delete duplicate values
addresses_df.drop_duplicates()

addresses_df.head()
```

```python
# save this file 
addresses_df.to_csv('cleaned_addresses.csv', index=False)

print('Saved file successfully....')
```

```python
# load customers.csv file

customers_df=pd.read_csv("C:/Users/Sistech Computer/OneDrive/Desktop/sandeep/enterprice-banking-analytics-system/data/raw/customers.csv")
customers_df.head()
```

```python
# converted DateOfBirth datatype from str to date
customers_df['DateOfBirth']=pd.to_datetime(customers_df['DateOfBirth'], format='mixed', errors='coerce')
```

```python
customers_df.head()
```
